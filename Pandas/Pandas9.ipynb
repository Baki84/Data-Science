```python
#range of dailty dates

pd.date_range(start = '2023/1/5' ,end= '2023/3/31' ,freq ='D')
```

```python
#alterantive days
pd.date_range(start = '2023/1/5' ,end= '2023/3/31' ,freq ='2D')
```

```python
#business days

pd.date_range(start = '2023/1/5' ,end= '2023/3/31' ,freq ='B')
```

```python
#one week per day

pd.date_range(start = '2023/1/5' ,end= '2023/3/31' ,freq ='W')
```

```python
#Month
pd.date_range(start='2026/09/14', end='2026/09/15', freq='ME')
```

```python
#using periods(no of result)

pd.date_range(start='2026/09/14', end='2026/09/15',periods=25)
```
