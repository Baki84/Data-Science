```python
#color and label
df = pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.1,color='red')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.1,color='silver')
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.1)

plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
```

```python
#label overlap

children = [10, 20, 40, 10, 30]
colors   = ['red', 'blue', 'green', 'yellow', 'pink']   # 5 labels, not 7

plt.bar(colors, children)
plt.xticks(rotation='vertical')
plt.tight_layout()   # prevents labels from being cut off
plt.show()
```

```python
#stacked bar chart
plt.bar(df['batsman'],df['2015'],color='red',label='2015')
plt.bar(df['batsman'],df['2016'],bottom=df['2015'],color='gold',label='2016')
plt.bar(df['batsman'],df['2017'],bottom=df['2015']+df['2016'],color='grey',label='2017')
plt.legend()
```
